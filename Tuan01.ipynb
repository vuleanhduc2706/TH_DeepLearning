```python
import numpy as np

print("--- CÂU 1 & 2 ---")
# 1 & 2. Tạo ma trận 3x3 ngẫu nhiên [0, 20].
arr_1_2 = np.random.randint(0, 21, size=(3, 3))
print("Ma trận:\n", arr_1_2)
# Kiểm tra tất cả khác 0 (hàm all) [cite: 573]
print("Tất cả phần tử khác 0:", np.all(arr_1_2 != 0))
# Kiểm tra tồn tại phần tử khác 0 (hàm any) [cite: 574]
print("Tồn tại phần tử khác 0:", np.any(arr_1_2 != 0))

print("\n--- CÂU 3 ---")
# 3. Dùng numpy.greater, greater_equal, less, less_equal so sánh 2 mảng. [cite: 574]
a1 = np.array([1, 5, 9])
a2 = np.array([5, 5, 5])
print("a1 > a2:", np.greater(a1, a2))
print("a1 >= a2:", np.greater_equal(a1, a2))
print("a1 < a2:", np.less(a1, a2))
print("a1 <= a2:", np.less_equal(a1, a2))

print("\n--- CÂU 4 ---")
# 4. Mảng gồm mười số 1, mười số 0 và mười số 5. [cite: 575]
arr_4 = np.concatenate([np.ones(10), np.zeros(10), np.full(10, 5)])
print(arr_4)

print("\n--- CÂU 5 ---")
# 5. Mảng số nguyên chẵn từ 30 đến 70. [cite: 576]
arr_5 = np.arange(30, 71, 2)
print(arr_5)

print("\n--- CÂU 6 ---")
# 6. Ma trận đơn vị 3x3. [cite: 577]
arr_6 = np.identity(3)
print(arr_6)

print("\n--- CÂU 7 ---")
# 7. Mảng 10 phần tử [15, 55], in các giá trị trừ phần tử đầu và cuối. [cite: 581, 582]
arr_7 = np.linspace(15, 55, 10)
print(arr_7[1:-1])

print("\n--- CÂU 8 ---")
# 8. Mảng 20 phần tử [0, 20], đổi dấu số nằm trong khoảng [9, 15]. [cite: 583]
arr_8 = np.arange(21)
arr_8[(arr_8 >= 9) & (arr_8 <= 15)] *= -1
print(arr_8)

print("\n--- CÂU 9 ---")
# 9. Ma trận 3x4 có giá trị [10, 21]. [cite: 584]
arr_9 = np.random.randint(10, 22, size=(3, 4))
print(arr_9)

print("\n--- CÂU 10 ---")
# 10. Ma trận 10x10, biên bằng 1, bên trong bằng 0. [cite: 585]
arr_10 = np.ones((10, 10))
arr_10[1:-1, 1:-1] = 0
print(arr_10)

print("\n--- CÂU 11 ---")
# 11. Ma trận 5x5 đường chéo chính là 1, 2, 3, 4, 5. [cite: 586]
arr_11 = np.diag([1, 2, 3, 4, 5])
print(arr_11)

print("\n--- CÂU 12 ---")
# 12. Mảng 3x3x3 tính tổng theo mỗi dòng, mỗi cột. [cite: 587]
arr_12 = np.random.randint(1, 10, size=(3, 3, 3))
print("Tổng theo cột (axis=0):\n", np.sum(arr_12, axis=0))
print("Tổng theo dòng (axis=1):\n", np.sum(arr_12, axis=1))

print("\n--- CÂU 13 ---")
# 13. 2 vector ngẫu nhiên 10 phần tử, tính inner product. [cite: 588]
v1 = np.random.rand(10)
v2 = np.random.rand(10)
print("Inner product:", np.dot(v1, v2))

print("\n--- CÂU 14 ---")
# 14. Ma trận A(4x3), vector y(3). Thêm y vào từng dòng của A. [cite: 589, 590]
A = np.random.randint(1, 10, size=(4, 3))
y = np.random.randint(1, 10, size=(3,))
print("A + y:\n", A + y)
```

```python
import tensorflow as tf

print("--- CÂU 1 ---")
# 1. Tạo tensor [2, 3] với các phần tử là 0
t1 = tf.zeros([2, 3])
print(t1)

print("\n--- CÂU 2 ---")
# 2. Tạo tensor giống kích thước X và gán 0
X = tf.constant([[1, 2, 3], [4, 5, 6]])
t2 = tf.zeros_like(X)
print(t2)

print("\n--- CÂU 3 ---")
# 3. Tạo tensor [3, 2] với các phần tử là 5
t3 = tf.fill([3, 2], 5)
print(t3)

print("\n--- CÂU 4 ---")
# 4. Tạo tensor 1-chiều với 50 phần tử chẵn trong khoảng [5, 10]
# Lưu ý: Vì khoảng [5, 10] chỉ có 3 số chẵn (6, 8, 10), nên ta sẽ random ngẫu nhiên
# 50 số từ tập {6, 8, 10} này.
t4 = tf.random.uniform([50], minval=3, maxval=6, dtype=tf.int32) * 2
print(t4)

print("\n--- CÂU 5 ---")
# 5. Tạo tensor 1-chiều [10, 12, 14, ..., 100]
t5 = tf.range(10, 101, delta=2)
print(t5)

print("\n--- CÂU 6 ---")
# 6. Tensor [3, 2] ngẫu nhiên (phân phối normal: mean=0, std=2)
t6 = tf.random.normal([3, 2], mean=0.0, stddev=2.0)
print(t6)

print("\n--- CÂU 7 ---")
# 7. Tensor [3, 2] ngẫu nhiên (phân phối uniform: [0, 2])
t7 = tf.random.uniform([3, 2], minval=0.0, maxval=2.0)
print(t7)

print("\n--- CÂU 8 ---")
# 8. Shuffle tensor X theo chiều thứ nhất
X8 = tf.constant([[1, 2], [3, 4], [5, 6]])
t8 = tf.random.shuffle(X8)
print(t8)

print("\n--- CÂU 9 ---")
# 9. Tạo tensor (10, 10, 3) và cắt ngẫu nhiên (random_crop) thành (5, 5, 3)
t9_goc = tf.random.normal([10, 10, 3])
t9_crop = tf.image.random_crop(t9_goc, size=[5, 5, 3])
print("Shape sau khi crop:", t9_crop.shape)

print("\n--- CÂU 10 ---")
# 10. Tạo ngẫu nhiên x, y (5 phần tử). Trả về x+y nếu x<y, ngược lại x-y
x_10 = tf.random.uniform([5], minval=0, maxval=10)
y_10 = tf.random.uniform([5], minval=0, maxval=10)
print("x:", x_10.numpy())
print("y:", y_10.numpy())
# Dùng hàm tf.where để check điều kiện cho từng cặp phần tử
t10 = tf.where(x_10 < y_10, x_10 + y_10, x_10 - y_10)
print("Kết quả:", t10.numpy())

print("\n--- CÂU 11 ---")
# 11. x = (3,2) trong [5, 10], y = (2,3) trong [15, 20]. Giữ đường chéo x và tính x+y
x_11 = tf.random.uniform([3, 2], minval=5.0, maxval=10.0)
y_11 = tf.random.uniform([2, 3], minval=15.0, maxval=20.0)
# Giữ đường chéo của x (biến các phần tử không thuộc đường chéo thành 0)
x_diag = tf.linalg.band_part(x_11, 0, 0)
# Vì x kích thước (3,2) còn y kích thước (2,3) không cộng trực tiếp được,
# ta cần ma trận chuyển vị của y (transpose) để cùng kích thước (3,2)
t11_result = x_diag + tf.transpose(y_11)
print(t11_result)

print("\n--- CÂU 12 ---")
# 12. Viết hàm toán học và đưa vào graph (@tf.function)
@tf.function
def graph_math_ops():
    x = tf.random.normal([3, 2, 2])
    y = tf.random.normal([3, 2, 2])
    return x+y, x-y, x*y, x/y

cong, tru, nhan, chia = graph_math_ops()
print("Kích thước sau khi cộng:", cong.shape)
```

```python
import tensorflow as tf

print("--- CÂU 13 ---")
class SumModel(tf.Module):
    @tf.function
    def __call__(self, x, y):
        return x + y

sum_model = SumModel()
print("Tổng:", sum_model(tf.constant(5), tf.constant(10)).numpy())

print("\n--- CÂU 14 ---")
class SubMulModel(tf.Module):
    @tf.function
    def __call__(self, x, y):
        return x - y, x * y

sub_mul_model = SubMulModel()
hieu, tich = sub_mul_model(tf.constant(10), tf.constant(5))
print(f"Hiệu: {hieu.numpy()}, Tích: {tich.numpy()}")

print("\n--- CÂU 15 ---")
# 15. Chu vi hình chữ nhật P = (a+b)*2
class PerimeterModel(tf.Module):
    @tf.function
    def __call__(self, a, b):
        return (a + b) * 2

chu_vi = PerimeterModel()
print("Chu vi HCN (a=4, b=3):", chu_vi(4, 3).numpy())

print("\n--- CÂU 16 ---")
# 16. Diện tích toàn phần hình hộp chữ nhật
class SurfaceAreaModel(tf.Module):
    @tf.function
    def __call__(self, a, b, c):
        s_xq = (a + b) * c * 2
        return (a * b) * 2 + s_xq

dt_toan_phan = SurfaceAreaModel()
print("Diện tích toàn phần (a=4, b=3, c=5):", dt_toan_phan(4, 3, 5).numpy())

print("\n--- CÂU 17 ---")
# 17. Chứng minh AD // BC <=> AD = BC và AB = CD
class TrapezoidModel1(tf.Module):
    @tf.function
    def __call__(self, AD, BC, AB, CD):
        return tf.logical_and(tf.equal(AD, BC), tf.equal(AB, CD))

cm_17 = TrapezoidModel1()
print("AD // BC là:", cm_17(AD=5, BC=5, AB=4, CD=4).numpy()) # Sẽ ra True

print("\n--- CÂU 18 ---")
# 18. Chứng minh AB = CD <=> AD = BC và AD // BC
class TrapezoidModel2(tf.Module):
    @tf.function
    def __call__(self, AD, BC, is_AD_parallel_BC):
        return tf.logical_and(tf.equal(AD, BC), is_AD_parallel_BC)

cm_18 = TrapezoidModel2()
print("AB = CD là:", cm_18(AD=5, BC=5, is_AD_parallel_BC=True).numpy())

print("\n--- CÂU 19 ---")
# 19. Chứng minh NA = NC <=> AM = MB và MN // BC
class TriangleMidsegmentModel(tf.Module):
    @tf.function
    def __call__(self, AM, MB, is_MN_parallel_BC):
        return tf.logical_and(tf.equal(AM, MB), is_MN_parallel_BC)

cm_19 = TriangleMidsegmentModel()
print("NA = NC là:", cm_19(AM=3, MB=3, is_MN_parallel_BC=True).numpy())
```
